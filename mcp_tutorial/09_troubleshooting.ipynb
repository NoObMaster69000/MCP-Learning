# Troubleshooting and Debugging

Common issues and how to solve them.

## 1. The 'Invisible Server' Problem

If your server doesn't show up in Claude Desktop:
1.  **Check JSON syntax**: Ensure `claude_desktop_config.json` is valid JSON.
2.  **Absolute Paths**: Always use absolute paths for the `command` and `args` in the config.
3.  **Environment**: Ensure the Python environment has `mcp` and `fastmcp` installed.

## 2. The 'Stdio Corruption' Problem

**NEVER** use `print()` for debugging in a stdio server. It will break the JSON-RPC communication.
- **Instead**: Use `logging` directed to `stderr`.

````python
import logging
import sys

# Correct logging setup
logging.basicConfig(
    level=logging.INFO,
    format="%(message)s",
    stream=sys.stderr
)
logging.info("This is safe and will appear in the client logs!")
````

## 3. Using the MCP Inspector

The Inspector is your best friend. It runs a web interface to test your server in isolation.

```bash
npx @modelcontextprotocol/inspector python your_server.py
```

---
**Summary**: Most MCP development issues can be solved by checking client logs, using absolute paths, and avoiding stdout for logging.
