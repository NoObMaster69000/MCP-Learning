## 1. Sampling (The Agentic Pattern)

Sampling allows a server to request a completion from the LLM via the client. This enables servers to become 'agentic' without needing their own API keys.

### Example Flow:
1. Server receives a request.
2. Server calls `session.create_message()` (Sampling).
3. Client forwards the request to the LLM.
4. LLM response is returned to the Server.

## 2. Roots and File System Security

Roots are a way for clients to tell servers which directories they are allowed to access. This is essential for safety when using file-handling tools.

```json
// Client-side root configuration
{
  "roots": [
    { "uri": "file:///home/user/projects/my-app", "name": "Project Root" }
  ]
}
```

## 3. Remote Servers with SSE

Server-Sent Events (SSE) allows for remote MCP connections. Unlike `stdio`, SSE uses HTTP and is suitable for cloud-deployed tools.

### SSE Lifecycle:
1. Client connects to `/sse` endpoint.
2. Server sends an `endpoint` event containing the POST URL for client messages.
3. Bidirectional communication begins over HTTP POST and SSE streams.
