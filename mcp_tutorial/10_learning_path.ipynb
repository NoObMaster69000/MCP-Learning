# Learning Path and Next Steps

Congratulations! You've completed the MCP tutorial. Where should you go from here?

## 1. Official Documentation

- [Model Context Protocol Home](https://modelcontextprotocol.io)
- [Python SDK GitHub](https://github.com/modelcontextprotocol/python-sdk)
- [MCP Specification](https://modelcontextprotocol.io/specification)

## 2. Open-Source MCP Projects

Explore the [MCP Server Registry](https://github.com/modelcontextprotocol/servers) to see what others have built (GitHub, Slack, Google Drive, etc.).

## 3. Suggested Exercises

- **Intermediate**: Build a server that queries a SQLite database.
- **Advanced**: Implement a multi-step agent workflow using **Sampling**.
- **Architect**: Design a remote MCP server using SSE and deploy it to a cloud provider.

---
**Summary**: MCP is a rapidly evolving standard. Stay involved with the community and keep building!
