# Setup and Installation

Let's get your environment ready for MCP development.

## 1. Prerequisites

- **Python 3.10+**
- `pip` or `uv` (highly recommended)

## 2. Environment Setup

It is best practice to use a virtual environment:

```bash
python -m venv .venv
source .venv/bin/activate  # On Windows: .venv\Scripts\activate
```

## 3. Installing Required SDKs

We will use the official Python SDK and `FastMCP`, which provides a high-level, decorator-based API.

```bash
pip install mcp fastmcp
```

## 4. Project Structure Best Practices

A typical MCP project looks like this:
```
my-mcp-server/
├── server.py        # Your server code
├── requirements.txt
├── .env             # For API keys
└── README.md
```

---
**Summary**: Setting up MCP is as simple as installing a few Python packages and organizing your project for easy maintenance.
