# MCP Best Practices

Guidelines for building reliable and maintainable MCP servers.

## 1. Descriptive Tool Metadata

Treat your tool descriptions as code. The LLM uses them to understand context.
- **Bad**: `def get_data(id): ...`
- **Good**: `def get_customer_record(customer_id: str): "Fetches detailed profile and purchase history for a specific customer ID." ...`

## 2. Progressive Disclosure

Don't overwhelm the LLM with too many tools in one server. If you have 50 tools, consider splitting them into multiple specialized servers.

## 3. Security: The Principle of Least Privilege

- Always use **Roots** to restrict file access.
- Never hardcode API keys; use environment variables.
- Validate all inputs. FastMCP uses Pydantic under the hood for type validation, but you should still check for logical errors.

## 4. Error Recovery

Return errors as text that the LLM can understand. 
- **Instead of**: `Exception("DB_ERR_01")`
- **Use**: `return "Error: Database connection timed out. Please try again in a few seconds."`

---
**Summary**: High-quality metadata, strict security, and human-readable error messages are the hallmarks of a professional MCP server.
