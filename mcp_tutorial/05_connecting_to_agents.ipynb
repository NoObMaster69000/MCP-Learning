# Connecting MCP with an AI Agent or LLM

Learn how to make your server available to Claude and other MCP-compatible clients.

## 1. Claude Desktop Integration

To use your server in Claude Desktop, you must configure the `claude_desktop_config.json` file.

### 📍 Location:
- **macOS**: `~/Library/Application Support/Claude/claude_desktop_config.json`
- **Windows**: `%APPDATA%\Claude\claude_desktop_config.json`

### 🔧 Configuration Example:

**Crucial**: Always use absolute paths for the `command` and `args`.

```json
{
  "mcpServers": {
    "knowledge-assistant": {
      "command": "python",
      "args": ["/ABSOLUTE/PATH/TO/mcp_tutorial/sample_server.py"]
    }
  }
}
```

## 2. Programmatic Client (Python)

Sometimes you want to build your own client to interact with an MCP server. Here's a robust way to do it using the SDK.

````python
import asyncio
import os
from mcp import ClientSession, StdioServerParameters
from mcp.client.stdio import stdio_client

async def run_client():
    # Get absolute path to the server script
    server_path = os.path.abspath("sample_server.py")
    
    server_params = StdioServerParameters(
        command="python",
        args=[server_path],
    )
    
    async with stdio_client(server_params) as (read, write):
        async with ClientSession(read, write) as session:
            await session.initialize()
            
            # List available tools
            tools = await session.list_tools()
            print(f"Found {len(tools.tools)} tools: {[t.name for t in tools.tools]}")

# asyncio.run(run_client())
````

## 3. The Discovery Process

When the client initializes, it calls `list_tools`, `list_resources`, and `list_prompts`. This 'handshake' is what allows the LLM to know exactly what capabilities your server provides without any manual configuration on the model's side.

---
**Summary**: Connecting a server to a client requires proper configuration (using absolute paths) or using the SDK to establish a session.
