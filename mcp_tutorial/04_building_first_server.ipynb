# Building Your First MCP Server

In this section, we will build a simple 'Math & Greeting' server using **FastMCP**.

## 1. Step-by-Step Implementation

We'll create a server that can add numbers and greet users.

````python
from fastmcp import FastMCP

# Create an MCP server named 'MyFirstServer'
mcp = FastMCP("MyFirstServer")

# Register a simple tool
@mcp.tool()
def add_numbers(a: int, b: int) -> int:
    """Adds two numbers together."""
    return a + b

# Register a tool with a description
@mcp.tool()
def greet(name: str) -> str:
    """Greets a person by name."""
    return f"Hello, {name}! Welcome to MCP."

# Register a resource
@mcp.resource("info://about")
def get_info() -> str:
    """Returns information about this server."""
    return "This is a demo MCP server built with FastMCP."
````

## 2. Explanation of Components

- `FastMCP("Name")`: Initializes the server instance.
- `@mcp.tool()`: A decorator that registers a Python function as an MCP tool. The docstring is used as the tool's description for the LLM.
- `@mcp.resource("uri")`: Registers a read-only data source accessible via a URI.

## 3. Running and Testing

You can run this server using the `mcp dev` command if you have the CLI installed, or simply run it as a script for debugging.

To test it locally, we use the **MCP Inspector**:
```bash
npx @modelcontextprotocol/inspector python server.py
```
