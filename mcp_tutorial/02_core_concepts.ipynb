# Core Concepts of MCP

Understanding the fundamental building blocks of MCP is crucial for designing effective architectures.

## 1. MCP Architecture Overview

MCP follows a **Client-Server model**:
- **Host**: The application where the user interacts with the AI (e.g., Claude Desktop, IDE).
- **Client**: The part of the host that maintains the connection to the server.
- **Server**: The specialized application that provides tools and data.

## 2. MCP Client vs MCP Server

- **Server**: Exposes capabilities. It says, 'I have these tools and this data.'
- **Client**: Consumes capabilities. It tells the LLM, 'Here are the tools available from this server.'

## 3. Primitives: Tools, Resources, and Prompts

### Tools (Executable)
Functions that the LLM can decide to run. They can have side effects (e.g., writing a file, sending an email).

### Resources (Read-only)
Data entities exposed by the server. Like URLs, they provide context (e.g., logs, documentation, database records).

### Prompts (Templates)
Reusable instruction templates that help standardize how the LLM should handle certain tasks.

## 4. Communication Flow and Lifecycle

1. **Initialization**: Client and Server agree on version and capabilities.
2. **Exchange**: The core phase where requests (e.g., call_tool) and responses happen.
3. **Termination**: A clean shutdown of the connection.

## 5. Transport Layers

- **stdio**: Standard Input/Output. Best for local servers. Simple and fast.
- **SSE**: Server-Sent Events. Best for remote servers over HTTP.

---
**Summary**: MCP uses a structured JSON-RPC communication over stdio or SSE to allow clients to discover and use tools, resources, and prompts from servers.
